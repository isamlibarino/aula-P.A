```python
matriz = [
    [1, 2],
    [3, 4]
]
soma = 0
for linha in matriz:
  for numero in linha:
    soma = soma + numero

print("Soma:", soma)
```

```python
matriz = [
    [8, 3, 12],
    [5, 20, 7]
]

maior = [0][0]
for linha in matriz:
  for numero in linha:
    if numero > maior:
      maior = numero

print("Maior numero:", maior)
```
