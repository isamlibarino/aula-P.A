Exercício 1 — Contagem

```python
numero = int(input("Digite um número: "))
contador = 1

while contador <= numero:
    print(contador)
    contador += 1
```

Exercício 2 — Contagem regressiva

```python
numero = int(input("Digite um número: "))
contador = numero

while contador >= 0:
    print(contador)
    contador -= 1
```

Exercício 3 — Números pares

```python
print("Números pares de 0 a 20:")
numero_par = 0

while numero_par <= 20:
    print(numero_par)
    numero_par += 2
```

Exercício 4 — Soma dos números

```python
numero = int(input("Digite um número: "))
soma = 0
contador = 1

while contador <= numero:
    soma += contador
    contador += 1

print(f"Soma: {soma}")
```

Exercício 5 — Tabuada

```python
numero = int(input("Digite um número: "))
mult = 1

print(f"Tabuada do {numero}:")
while mult <= 10:
    resultado = numero * mult
    print(f"{numero} x {mult} = {resultado}")
    mult += 1
```

Exercício 6 — Senha

```python
senha_correta = "1234"
senha_digitada = ""

while senha_digitada != senha_correta:
    senha_digitada = input("Digite a senha: ").strip()
    if senha_digitada != senha_correta:
        print("Senha incorreta!")

print("Acesso permitido!")
```

Exercício 7 — Somando até digitar zero

```python
soma = 0
numero = int(input("Digite um número: "))

while numero != 0:
    soma += numero
    numero = int(input("Digite um número: "))

print(f"Soma: {soma}")
```

Exercício 8 — Maior número

```python
maior = 0
numero = int(input("Digite um número: "))

while numero != 0:
    if numero > maior:
        maior = numero
    numero = int(input("Digite um número: "))

print(f"Maior número: {maior}")
```

Exercício 9 — Média das notas

```python
soma = 0
contador = 1

while contador <= 5:
    nota = float(input(f"Digite a nota {contador}: "))
    soma += nota
    contador += 1

media = soma / 5
print(f"Média: {media}")
```

Exercício 10 — Caixa eletrônico

```python
saldo = 1000
opcao = 0

while opcao != 4:
    print("Escolha uma opção:")
    print("1 - Saldo")
    print("2 - Depósito")
    print("3 - Saque")
    print("4 - Sair")

    opcao = int(input("Digite a opção desejada: "))
    if opcao == 1:
        print(f"Saldo: R${saldo}")
    elif opcao == 2:
        deposito = float(input("Digite o valor do depósito: "))
        saldo += deposito
    elif opcao == 3:
        saque = float(input("Digite o valor do saque: "))
        if saque > saldo:
            print("Saldo insuficiente!")
    elif opcao == 4:
        print("Saindo")
    else:
        print("Opção inválida!")
```
