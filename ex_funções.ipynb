```python
def verificar_numero(numero):
  if numero > 0:
    print("O número é positivo")
  elif numero < 0:
    print("O número é negativo")
  else:
    print("O número é zero")
verificar_numero(10)
verificar_numero(-7)
verificar_numero(0)
```

```python
def calcular_media(nota1,nota2):
  media = (nota1 + nota2) / 2
  print("A média é",media)
calcular_media(9,6)
```
