```python
nome = input("Digite o seu nome: ")
nome = nome.upper()
print(f"Seu nome em maiúsculo é: {nome}")
```

```python
cidade = input("Digite uma cidade: ")
cidade = cidade.lower()
print(f"O nome da cidade em minúsculo é: {cidade}")
```
