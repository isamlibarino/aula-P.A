EXERCÍCIOS MATCH

Ex 1


```python
cor = "azul"

match cor:
    case 1:
        print("Cor azul")
    case _:
        print("Outra cor")
```

Ex 2

```python
numero = 2

match numero:
    case 1:
        print("Um")
    case 2:
        print("Dois")
    case 3:
        print("Três")
    case _:
        print("Número não cadastrado")
```

Ex 3

```python
opcao = 1

match opcao:
    case 1:
        print("Água")
    case 2:
        print("Suco")
    case 3:
        print("Refrigerante")
    case _:
        print("Opção inválida")
```

Ex 4

```python
dia = "sexta-feira"

match dia:
    case "sábado":
        print("Final de semana")
    case "domingo":
        print("Final de semana")
    case _:
        print("Dia de semana")
```

Ex 5

```python
nota = "C"

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```

EXERCÍCIOS RANGE

Ex 1

```python
for numero in range(10):
  print(numero)
```

Ex 7

```python
tabuada = 3
for i in range(1, 11):
    resultado = tabuada * i
    print(f"{tabuada} x {i} = {resultado}")
```
